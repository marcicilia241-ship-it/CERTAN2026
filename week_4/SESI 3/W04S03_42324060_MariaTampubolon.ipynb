Naïve Bayes

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

pada kode ini merupakan awal dalam memuat dataset yang memiliki fungsi untuk mengimpor pustaka utama seperti numpy dan pandas untuk pengolahan data, serta memuat dataset baawan yang terkenal di dalam machine learning.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

```python
Di tahap ini yaitu data mentah dari dataset iris diubah dalam bentuk nama spesies ditambhkan, 
lalu data dibagi 2 bagian
```

```python
y
```

baris ini digunakan untuk menampilkan isi dari variabel target (y), yang berisikan nama spesies bunga untutk setiap baris data 

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

Berikut merupakan Model machine learning lebih cepat memproses angka dari pada teks. 
LabelEncoder digunakan untuk mengubah nama spesie(teks) menjadi angka.

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Pada code diatas merupakan dataset yang dibagi dua bagian:data untuk melatih model(training set) dan data untuk menguji model(testing set) dengan perbandingan 80% untutk latihan dan 20% untuk menguji menggunakan train_test_split.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

Kode diatas karena semua fitur pada dataset Iris berupa kontinu(desimal), algoritma yang dipilih adalah Gaussian Naïve Bayes, Model kemudian dilatih menggunakan data latih

```python
result = NB_model.predict(X_test)
result
```

Berikut melakukan prediksi pada data uji, Model yang sudah dilatih dengan pintar dari hasil pelatihan tadi digunakan untuk menebak spesies bunga pada data pengujian. Hasil prediksinya berupa deretan angka kategori spesies.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

Kode ini menghitung tingkat akurasi dengan membandingkan hasil tebakan model dengan data asli yang sebenarnya (y_test) menggunakan accuracy_score.

```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Berikut percobaan prediksi data baru secara langsung menggunakan array angka. Munculnya peringatan warning karena model sebelumnya dilatih menggunakan DataFrame yang memiliki nama kolom, sementara input baru dimasukkan sebagai array biasa tanpa kolom

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Code diatas merupakan solusi memperbaiki peringatan data baru seperti output yang sebelumnya, untuk menghilangkan peringatan diatas, input data baru diubah terlebih dahulu menjadi format DataFrame Pandas lengkap dengan nama kolon tanpa yang sesuai. Hasilnya , prediksi berlajalan lancar tanpa adanya peringatan seperti yang sebelumnya.
