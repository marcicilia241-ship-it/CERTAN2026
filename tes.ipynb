```python
# operasi elementwise (mengoperasikan setiap elemen yang berkorespondensi dengan elemen yang sama)
a = np.array([1, 2, 3, 4, 5])
b = np.array([6, 7, 8, 9, 10])

hasil1 = a+b
hasil2 = a-b
hasil3 = a*b
hasil4 = a/b
hasil5 = a**2

print("hasil 1 =", hasil1)
print("hasil 2 =", hasil2)
print("hasil 3 =", hasil3)
print("hasil 4 =", hasil4)
print("hasil 5 =", hasil5)
```

```python
# operator Elementwise ada masalah di perkalian matriks karena mengalikan per elemen
a = np.array( [(1,2,3), (4,5,6)] )
b = np.array( [(3,4,5), (3,4,5)] )

hasil1 = a*b

print("hasil1 =", hasil1)
```

```python
#perkalian matriks
a = np.array( [(1,2,3),(4,5,6)] )
b = np.array( [(1,2,3),(1,2,3),(1,2,3)] )
hasil_perkalian_matriks = np.dot(a,b)
print(hasil_perkalian_matriks)
```

```python
#indexing
a = np.arange(10)**3
print(a)

print("mengambil elemen ke 1 dari a =", a[0])
print("mengambil elemen ke 4 dari a =", a[3])
print("mengambil elemen terakhir dari a =", a[-1])
print("mengambil elemen ke 1 dari a =", a[0])

print("mengambil elemen dari 1 sampai 4 =", a[0:4])
print("mengambil elemen dari 4 sampai terakhir =", a[3:])
print("mengambil elemen dari awal sampai 4 =", a[:4])
```

```python
#mengetahui ukuran matriks
a = np.array( [(1,2,3),(4,5,6)] )
print("matriks a berukuran =", a.shape)

#transpose matriks
print("transpose matriks a =", a.transpose())
print("transpose matriks a =", np.transpose(a))
print("transpose matriks a =", a.T)

#Pada Machine learning biasanya flatten dan stacking sangat penting

#flatten array, vektor baris
print("flatten dari matriks a =", a.ravel())
```

```python
#Pada Machine learning biasanya flatten dan stacking sangat penting
#stacking array atau menumpuk array
a = np.array([1,2,3])
b = np.array([4,5,6])

a1 = np.hstack((a,b))
a2 = np.vstack((a,b))

print("horizontal stack =", a1)
print("vertikal stack =", a2)
```

```python
#invers matriks
a = np.array( [(1,-1), (1,1)] )
print(a)

a_inv = np.linalg.inv(a)
print(a_inv)

det_a = np.linalg.det(a)
print(det_a)
```

```python
#solusi persamaan linear
A = np.array( [(1,-6),(2,-5)] )
Y = np.array([-35, 21])

print(A)
print(Y)

A_inv = np.linalg.inv(A)

#menyelesaikan persamaan linear
X1 = np.dot(A_inv, Y)
print(X1)

#cara lain
X2 = np.linalg.solve(A,Y)
print(X2)
```

```python
def f(x,y):
    return np.cos(np.exp(-x**2-y**2))*np.sin(np.sqrt(x**2+y**2))

x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)

X,Y = np.meshgrid(x,y)

Z = f(X,Y)

fig = plt.figure()

ax = plt.axes(projection = '3d')

ax.contour3D(X, Y, Z, 50, cmap = 'binary')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
```

```python
# SOAL NO 2
# Grafik 2 dimensi

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 400)
y = np.linspace(-2, 2, 400)

X, Y = np.meshgrid(x, y)

Z = (X**6
     + 3*X**4*Y**2
     - 3*X**4
     + 3*X**2*Y**4
     - X**2*Y**3
     - 6*X**2*Y**2
     + 3*X**2
     + Y**6
     - 3*Y**4
     + 3*Y**2)

plt.contour(X, Y, Z, levels=[1])

plt.xlabel('x')
plt.ylabel('y')
plt.title('Grafik 2 Dimensi')

plt.grid()
plt.show()
```

```python
# SOAL NO 3
# Grafik 3 dimensi

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z = ((5*Y/4) - np.abs(X))**2 + X**2

fig = plt.figure()

ax = plt.axes(projection='3d')

ax.contour3D(X, Y, Z, 50)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

ax.set_title('Grafik 3 Dimensi')

plt.show()
```
